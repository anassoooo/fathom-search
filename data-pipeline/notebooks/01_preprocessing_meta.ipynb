```python
def floatPrice(s: pd.Series) -> pd.Series:
    return (
        s.astype(str)
         .str.replace("$", "", regex=False)
         .str.replace(",", "", regex=False)
         .astype(float)
    )
data['Product Price ($)']=floatPrice(data['Product Price'])
```

```python
import os
import mimetypes
import pandas as pd
import requests

OUT_DIR  = "images"               # change
ID_COL   = "Product ID"           # change if needed
URL_COL  = "Product Image"        # change if needed

os.makedirs(OUT_DIR, exist_ok=True)


session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/avif,image/webp,image/apng,image/*,*/*;q=0.8",
})

def guess_ext(resp, url):
    ct = resp.headers.get("Content-Type", "").split(";")[0].strip()
    ext = mimetypes.guess_extension(ct) if ct else None
    if ext:
        return ext
    # fallback from URL
    tail = url.split("?")[0].lower()
    for e in [".jpg", ".jpeg", ".png", ".webp"]:
        if tail.endswith(e):
            return e
    return ".jpg"

ok, fail = 0, 0
for _, row in data.iterrows():
    pid = str(row[ID_COL])
    url = str(row[URL_COL]) if pd.notna(row[URL_COL]) else ""
    if not url or url == "nan":
        fail += 1
        continue

    try:
        r = session.get(url, stream=True, timeout=(5, 30))
        r.raise_for_status()

        ext = guess_ext(r, url)
        path = os.path.join(OUT_DIR, f"{pid}{ext}")

        with open(path, "wb") as f:
            for chunk in r.iter_content(chunk_size=1024 * 64):
                if chunk:
                    f.write(chunk)

        ok += 1
    except Exception:
        fail += 1

print(f"Downloaded: {ok} | Failed: {fail}")
```

```python
def intTotalRatings(s: pd.Series) -> pd.Series:
    return (
        s.astype(str)
         .str.replace(",", "", regex=False)
         .astype(int)
    )
data['Total Ratings']=intTotalRatings(data['Total Ratings'])
data.head()
```

```python
data[data['Product Price ($)']==0]['Product Subcategory'].value_counts()
```

```python
# treat 0 as missing
import numpy as np
data.loc[data['Product Price ($)'] <= 0, 'Product Price ($)'] = np.nan

# compute median per subcategory and fill
sub_median = data.groupby('Product Subcategory')['Product Price ($)'].transform("median")
data['Product Price ($)'] = data['Product Price ($)'].fillna(sub_median)
```
